# 🧬 TCGA Breast Cancer Mutation Analysis

### 🎯 Objective
This project analyzes somatic mutation data from TCGA breast cancer samples to:
- Identify the most frequently mutated genes
- Explore mutation types across patients
- Break down mutation classifications by gene

```python
import pandas as pd

# Load MAF file, skip comment lines and problematic rows
df = pd.read_csv("data_mutations.txt", sep="\t", comment='#', on_bad_lines='skip', low_memory=False)

# Preview the data
df.head()
```

## 📊 Column Overview
Let's check which columns are available in this MAF file.

```python
print(df.columns.tolist())
```

## 🔬 Top Mutated Genes
We'll count how often each gene appears in the dataset and visualize the top 10 most frequently mutated ones.

```python
mutation_counts = df['Hugo_Symbol'].value_counts()

print("Top 10 most mutated genes:")
print(mutation_counts.head(10))
```

```python
import matplotlib.pyplot as plt

mutation_counts.head(10).plot(kind='bar', title='Top 10 Mutated Genes')
plt.xlabel('Gene')
plt.ylabel('Mutation Count')
plt.tight_layout()
plt.show()
```

## 👤 Top Mutated Samples
Now we'll see which tumor samples (patients) have the most mutations.

```python
mutations_per_sample = df['Tumor_Sample_Barcode'].value_counts()

print("Top 10 samples with the most mutations:")
print(mutations_per_sample.head(10))
```

```python
mutations_per_sample.head(10).plot(kind='bar', title='Top 10 Mutated Samples')
plt.xlabel('Sample')
plt.ylabel('Mutation Count')
plt.tight_layout()
plt.show()
```

## 🔁 Mutation Type Breakdown for Top Genes
For the top 5 most mutated genes, we'll show how many different types of mutations (e.g., missense, nonsense) occurred.

```python
# Find top 5 most frequently mutated genes
top_genes = df['Hugo_Symbol'].value_counts().head(5).index.tolist()

# Subset the dataframe to include only those top genes
df_top_genes = df[df['Hugo_Symbol'].isin(top_genes)]

# Create pivot table: gene on rows, mutation types as columns
gene_mutation_pivot = df_top_genes.pivot_table(
    index='Hugo_Symbol',
    columns='Variant_Classification',
    aggfunc='size',
    fill_value=0
)

# Show the breakdown table
gene_mutation_pivot
```

```python
gene_mutation_pivot.plot(kind='bar', stacked=True, figsize=(10,6), title='Mutation Types in Top 5 Genes')
plt.xlabel('Gene')
plt.ylabel('Mutation Count')
plt.tight_layout()
plt.show()
```

## ✅ Summary

- **TP53** and **PIK3CA** were among the most frequently mutated genes in the breast cancer dataset.
- Certain samples show unusually high mutation counts, indicating high mutation burden.
- Mutation types vary by gene, with TP53 showing diverse types like missense and nonsense mutations.
